```python
# B
import pandas as pd

data = {
'Order_ID':['O1','O2','O3','O4','O5','O6','O7','O8','O9','O10'],

'Category':['Electronics','Clothing','Electronics','Grocery','Clothing',
            'Electronics','Grocery','Clothing','Electronics','Grocery'],

'Payment_Method':['UPI','Card','UPI','Cash','Card',
                  'UPI','Cash','Card','UPI','Cash'],

'Delivery_Type':['Express','Standard','Express','Standard','Express',
                 'Standard','Express','Standard','Express','Standard'],

'Customer_Type':['New','Returning','Returning','New','Returning',
                 'New','Returning','New','Returning','New']
}
```

```python
df = pd.DataFrame(data)
print(df)
```

```python
#1. frequency count
print(df['Category'].value_counts())
```

```python
#2. percentage pf each element
print(df['Category'].value_counts(normalize=True)*100)
```

```python
#3. payment
print(df.Payment_Method.value_counts())
```

```python
#4. unique values
print(df["Category"].unique())
```

```python
#5. number of unique values
print(df["Category"].nunique())
```

```python
#6. cross tabulation
print(pd.crosstab(df['Category'], df['Payment_Method']))
```

```python
#7. filtering data : orders from electronics category
electronics_order = df[df['Category'] == 'Electronics']
print(electronics_order)
clothing_order = df[df['Category'] == 'Clothing']
print(clothing_order)
groceries_order = df[df['Category'] == 'Grocery']
print(groceries_order)
```

```python
#8. sorting data
sorted_data1 = df.sort_values(by='Category')
print(sorted_data1)
```

```python
#9. grouping
#orders by category and payment method
print(df.groupby('Category')['Payment_Method'].value_counts())
```
